Let's write a Python program to:
1. Generate 1,00,000 random numbers.
2. Compute mean, median, mode in two ways:
A. From scratch (without NumPy built-ins)
B. Using NumPy built-ins
3. Compare execution time of both methods.
we can implement all basic descriptive statistics using NumPy (and sometimes SciPy for advanced ones).

```python
import numpy as np
import numpy.random as ran
import time
from scipy import stats
from collections import Counter

data=[ran.randint(1,1000) for _ in range(1000000)]

np_data=np.array(data)
```

```python
def mean_from_scratch(arr):
    return sum(arr)/len(arr)

def median_from_scratch(arr):
    sorted_array=sorted(arr)
    n=len(sorted_array)
    mid=n//2
    if n%2==0:
        return (arr[mid]+arr[mid-1])/2
    else:
        return arr[mid]
        
def mode_from_scratch(arr):
    freq = Counter(arr)
    max_freq = max(freq.values())
    modes = [k for k, v in freq.items() if v == max_freq]
    
    return modes[0] if len(modes) == 1 else modes

mode_from_scratch(np_data)
```

```python
start=time.time()
means_s=mean_from_scratch(np_data)
median_s=median_from_scratch(np_data)
mode_s=mode_from_scratch(np_data)
end=time.time()

scratch_time=end-start
scratch_time
```

```python
start=time.time()
mean_np=np.mean(np_data)
median_np=np.median(np_data)
mode_np=stats.mode(np_data,keepdims=True)
end=time.time()

numpy_time=end-start
numpy_time
```

Bar Plot

```python
df.columns
```

```python
Categories_labels=['bedrooms', 'bathrooms', 'stories', 'mainroad',
       'guestroom', 'basement', 'hotwaterheating', 'airconditioning',
       'parking', 'prefarea', 'furnishingstatus']
for i in Categories_labels:
    plt.figure(figsize=(8,3))
    sns.countplot(x=i,data=df)
    plt.title(f'Distribution of {i} data')
    plt.show()
```

Seaborn

```python
plt.figure(figsize=(8,3))
sns.barplot(x=df["area"])
plt.title("Box Plot of Area",fontsize=14)
plt.xlabel('Area (sq ft)',fontsize=12)
plt.show() 
df['area']
```

```python
df['area'].plot.box()
```

```python
plt.figure(figsize=(8,4))
sns.kdeplot(df['area'],fill=True,color='blue',alpha=0.4,linewidth=2)
plt.title("KDE plot of Area",fontsize=14)
plt.xlabel("Area (sq ft)")
plt.ylabel("Density",fontsize=12)
plt.show()
```

```python
plt.figure(figsize=(8,3))
sns.boxplot(x=df['price'])
plt.title("Box plot of Price")
plt.xlabel("Price" ,fontsize=12)
plt.show()
```

Bivariate Analysis

```python
numeric_cols=['area']

for col in numeric_cols:
    plt.figure(figsize=(10,5))
    sns.scatterplot(x=col,y="price",data=df,alpha=0.6)
    plt.title(f"Price vs {col}",fontsize=14)
    plt.xlabel(col,fontsize=12)
    plt.ylabel("Price",fontsize=12)
    plt.show()
```

```python
for col in Categories_labels:
    plt.figure(figsize=(10,4))
    sns.boxplot(x=col,y="price",data=df)
    plt.xlabel(col,fontsize=12)
    plt.ylabel("Price",fontsize=12)
    plt.show()
```

```python
NumericalColumns = ["price", "area", "bedrooms", "bathrooms", "stories", "parking"]
plt.figure(figsize=(10,8))
correlation_m=df[NumericalColumns].astype(float).corr()
sns.heatmap(correlation_m,cmap='coolwarm',vmin=-1,vmax=1,annot=True)
```
